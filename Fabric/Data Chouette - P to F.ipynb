# Large Semantic Model > 10 GB

```python
df_models = labs_admin.list_datasets()
df_models_filtered = df_models[df_models['Dataset Id']=='107fd8fc-4ae1-4daa-a913-1229e535e758']
df_models_filtered
```

## 1 - Assign Workspaces to Storage Account

```python
import sempy_labs as labs
labs.assign_workspace_to_dataflow_storage(dataflow_storage_account='bckstore', workspace='DataChouette1-Large10')
```

## 2 - Backup Semantic Model

```python
labs.backup_semantic_model(
    dataset='107fd8fc-4ae1-4daa-a913-1229e535e758',
    file_path=f'Sales.abf',
    allow_overwrite=True, 
    apply_compression=True,
    workspace='DataChouette1-Large10'
)
```

## 3 - Process Clear to empty Semantic Model

```python
labs.refresh_semantic_model(dataset='107fd8fc-4ae1-4daa-a913-1229e535e758', refresh_type='clearValues', workspace='DataChouette1-Large10')
```

```python
labs.vertipaq_analyzer(dataset='Sales', workspace='DataChouette1-Large10')
```

## 4 - Set Storage Format to Small

```python
labs.set_semantic_model_storage_format(dataset='Sales', workspace='DataChouette1-Large10',storage_format='Small')
```

## 5 - Migrate to F Capacity

```python
admin.assign_workspaces_to_capacity(
    source_capacity='datachouette1',
    target_capacity='datachouette2',
    workspace='DataChouette1-Large10'
)
```

## 6 - Set Storage Format to Large

```python
labs.set_semantic_model_storage_format(dataset='Sales', workspace='DataChouette1-Large10',storage_format='Large')
```

## 7 - Restore Semantic Model

```python
labs.restore_semantic_model(
    dataset='Sales',
    file_path=f'Sales.abf',
    allow_overwrite=True,
    ignore_incompatibilities=True,
    workspace='DataChouette1-Large10',
    force_restore=True
)
```

## 8 - Verify

```python
labs.vertipaq_analyzer(dataset='Sales', workspace='DataChouette1-Large10')
```
